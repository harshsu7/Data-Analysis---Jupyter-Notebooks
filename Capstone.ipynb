```python
df.replace('?',np.nan, inplace = True)
df.isnull().sum()
```

```python
df['price'] = df['price'].astype(float)
```

```python
df["price"] = pd.to_numeric(df["price"], errors="coerce")
df["num-of-doors"].fillna(df["num-of-doors"].mode()[0], inplace=True)
```

```python
for col in ["price", "horsepower", "bore", "stroke", "peak-rpm"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
# Fill missing numeric values with median
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())
```

```python
categorical_cols = df.select_dtypes(include=["object"]).columns
print(categorical_cols)
```

```python
# One-hot encode categorical features
df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True)

print(df_encoded.head())
```

```python
corr_matrix = df_encoded.corr()

plt.figure(figsize=(15,10))
sns.heatmap(corr_matrix, cmap="coolwarm", center=0)
plt.title("Correlation Heatmap (Including Encoded Categorical Features)")
plt.show()
```

```python
price_corr = corr_matrix["price"].sort_values(ascending=False)
print(price_corr.head(15))   # Top positive correlations
print(price_corr.tail(15))   # Top negative correlations
```

```python
for col in categorical_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```
