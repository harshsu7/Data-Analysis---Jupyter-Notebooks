# Seaborn from Zero to Advanced
Using Mumbai House Prices Dataset

This notebook teaches Seaborn **from basics to advanced**, with *why* each plot is used.

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\NewUser\Downloads\Mumbai House Prices.csv")
df.head()
```

## Why Seaborn
Seaborn is used for statistical visualization. It works directly with DataFrames.

```python
sns.set_style('whitegrid')
sns.histplot(df['price'], kde=True)
plt.show()
sns.despine(right = True)
```

Histogram shows distribution. KDE shows smooth density.

```python
sns.set_style('dark')
sns.kdeplot(df['price'], fill=True)
plt.show()
```

```python
sns.set_style('white')
sns.boxplot(y=df['price'])
plt.show()
```

```python
df.columns
```

```python
sns.countplot(x='region', data=df)
plt.xticks(rotation=45)
plt.show()
```

```python
sns.stripplot(x='area', y='price',hue = 'bhk', palette='Dark2' ,data=df)
plt.show()
```

```python
sns.set_style('darkgrid')
sns.scatterplot(x='area', y='price',hue = 'bhk', palette='Dark2' ,data=df,alpha=0.6)
plt.show()
```

```python
sns.color_palette('Set2')
```

```python
sns.color_palette('Set3')
```

```python
sns.set_style('ticks')
sns.scatterplot(x='area', y='price',hue='region',data=df)
plt.show()
```

```python
sns.regplot(x='area', y='price', data=df)
plt.show()
sns.set_context('talk')
sns
```

```python

sns.violinplot(y='price', data=df)
plt.show()
```

```python

corr = df.select_dtypes(include='number').corr()
sns.heatmap(corr, annot=True)
plt.show()
```

```python

sns.pairplot(df.select_dtypes(include='number'))
plt.show()
```

```python
sns.swarmplot(x='area', y='price',hue = 'bhk', palette='Dark2' ,data=df)
plt.show()
```
