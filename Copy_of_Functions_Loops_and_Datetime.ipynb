```python
def calculate(operation, *numbers):
    if not numbers:
        return "No numbers provided!"

    result = numbers[0]

    if operation == '+':
        return sum(numbers)

    elif operation == '*':
        result = 1
        for n in numbers:
            result *= n
        return result

    elif operation == '-':
        for n in numbers[1:]:
            result -= n
        return result

    elif operation == '/':
        for n in numbers[1:]:
            if n == 0:
                return "Error: Division by zero!"
            result /= n
        return result

    else:
        return "Invalid operation!"


while True:
    op = input("\nEnter operation or 'q' to quit: ")
    if op == 'q':
        break


    nums_input = input("Enter numbers separated by spaces: ")
    num_list = [float(x) for x in nums_input.split()]

    ans = calculate(op, *num_list)
    print(f"Result: {ans}")
```

```python
from datetime import datetime, timedelta

def check_subscription(expiry_date_str):

    expiry_date = datetime.strptime(expiry_date_str, "%Y-%m-%d")
    today = datetime.now()

    remaining_time = expiry_date - today

    if remaining_time.days < 0:
        return "Expired"
    elif remaining_time.days <= 7:
        return f"Urgent: {remaining_time.days} days left!"
    else:
        readable_date = expiry_date.strftime("%d %B, %Y")
        return f"Active until {readable_date} ({remaining_time.days} days remaining)"


user_input = input("Enter expiry date (YYYY-MM-DD): ")
status = check_subscription(user_input)
print(f"Subscription Status: {status}")
```

# OOPS
- Classes are blueprints (home blueprint) and objects are the real home with bricks and cements

- objects can be termed as bundles of related attributes("variables") and methods("Functions")


```python
class car:  #defining a class
  car_type = "Manual"

  def __init__(self,model,color,year):  # defining a constructor
    self.model = model
    self.year = year
    self.color = color

  def drive(self):
    print(f"you are driving the {self.color} {self.model}")
```

# Class Variable and Instance Variable

- Instance variables are unique to that specific object. Class variables are 'global' to the whole class. If the recipe changes, all future cakes are affected

```python
class Cake:
    bakery_name = "The Best Bakery" # Class Variable (shared)

    def __init__(self, flavor):
        self.flavor = flavor # Instance variable (unique)

print(Cake.bakery_name) # "The Best Bakery"

cake1 = Cake("Chocolate")
print(cake1.bakery_name)
```

# Encapsulation

- We use underscores to tell other developers how to treat our data
1. No underscore: Public (It's a free-for-all)
2. _: Protected (Warning: Keep out).
3. __: Private (I’ve hidden it so well even the compiler gets confused)

```python
class DataBox:
    def __init__(self):
        self.public = "I am visible"
        self._protected = "Caution"
        self.__private = "Secret" # Hidden

box = DataBox()
print(box.public)
print(box._protected)
#print(box.__private) # This would throw an error!
```

# Getters and Setters (The "Security Guard")
- We hide the data to keep it safe, but we still need a way to read or change it. We use methods as 'Security Guards' to check if the new data is valid before allowing it in.

```python
class UserProfile:
    def __init__(self, age):
        self.__age = age # Private

    @property
    def age(self): # Getter
        return self.__age

    @age.setter
    def age(self, value): # Setter
        if value < 0:
            print("Age cannot be negative!")
        else:
            self.__age = value

user = UserProfile(25)
user.age = -5 # Triggering the setter!
print(user.age) # Triggering the getter!
```

```python
class FinancialRecord:
    Position = "CA"
    def __init__(self, value):
        self._value = value # _ indicates it's private

    @property
    def value(self): # Getter - to get a value from a secret variable
        return self._value

    @value.setter
    def value(self, new_val): # Setter - to set a new value for a secret variable
        if new_val < 0:
            print("Error: Salary cannot be negative!")
        else:
            self._value = new_val
```

```python
f1 = FinancialRecord(1000)
print(f1.value)
print(f1.Position)
f1.value = -500
print(f1.value)
```

```python
class DataCleaner:
    def __init__(self, data):
        self.data = data  # Attribute

    def remove_nulls(self):
        return [x for x in self.data if x is not None]

# Usage
my_data = DataCleaner([1, 2, None, 4])
print(my_data.remove_nulls())
```
