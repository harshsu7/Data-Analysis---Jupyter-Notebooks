```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
x = np.random.rand(100)
y = np.random.rand(100)
df = pd.DataFrame({'col1':x, 'col2':y})
```

```python
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr"],
    "Sales": [200, 250, 300, 280],
    "Region": ["East", "West", "East", "West"]
}
df = pd.DataFrame(data)
```

```python
plt.plot(df["Month"], df["Sales"], color="green", marker="o")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.show()
```

```python
sns.lineplot(x="Month", y="Sales", data=df)
plt.show()
```

```python
sns.lineplot(x='Month', y='Sales',marker='o',data=df)
plt.title('Random Scatter Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
plt.bar(df["Month"], df["Sales"])
plt.show()
```

```python
sns.barplot(x="Month", y="Sales", data=df)
plt.show()
```

```python
sns.barplot(x="Month", y="Sales", hue="Region", data=df)  #Seaborn understands data relationships 
plt.show()
```

```python
sns.histplot(df["Sales"], kde=True)  #Only Available in Seaborn - Used to understand spread, not totals
plt.show()
```

```python
sns.kdeplot(df["Sales"], fill=True)  #Only Available in Seaborn - Used to understand spread, not totals
```

```python
sns.boxplot(x="Region", y="Sales", data=df, hue = "Region", orient= None, showfliers = True)  #Only Available in Seaborn - Used to understand spread, not totals
```

```python
#The box Contains the middle 50% of data, From Q1 to Q3, Box size = spread
#line in box = median , Where half the data is below, half above
# Median closer to bottom then data is rightly skewed
# Median closer to top then left skewed
# Show how far data extends normally - Not max/min always, just non-outlier range
# Big box → data is scattered
# Small box → data is tight
```

```python
sns.boxenplot(x="Region", y="Sales", data=df)  #Enhanced Box Plot for larger datasets
```
